# **Loss given default (LGD) modeling**

This notebook presents a methodology for building a model to estimate LGD (Loss Given Default).We work with the dataset that has already been cleaned and preprocessed in the earlier steps.

LGD measures the proportion of exposure that is lost when a borrower defaults. It quantifies the financial loss severity, complementing default probability estimates for a full credit risk assessment.

## **Defining the target variable**

When building LGD models, it is best to use data from borrowers who have had sufficient time to repay part of their debt. Therefore, we will focus only on written-off loans, indicated by loan_status values:

- 'Charged Off'

- 'Does not meet the credit policy. Status: Charged Off'

From the available data:

- `funded_amnt` represents the total amount lost at the moment of default.

- `recoveries` represents the amount recovered.

Thus, LGD can be calculated as:


$LGD = \frac{funded\_amnt - recoveries}{funded\_amnt} = 1 - recovery\_rate$


For this analysis, the focus will be on modeling the recovery rate. Once the recovery rate is predicted, LGD is simply computed as 1 – recovery rate.

```python
default_categories = [
    "Charged Off",
    "Does not meet the credit policy. Status:Charged Off"
]

df = df[df["loan_status"].isin(default_categories)]
```

```python
target_variable = "RR"

df[target_variable] = df["recoveries"] / df["funded_amnt"]
df[target_variable].describe()
```

Since the recovery rate must lie between 0 and 1, we clip the values to ensure they stay within this range:

```python
df[target_variable] = np.clip(df[target_variable], a_min=0.0, a_max=1.0)
```

## **Modeling**

As observed earlier, the distribution of the recovery rate (RR) is highly skewed, with many values clustered at zero. A standard regression model would struggle to predict these zeros accurately. To address this, a Zero-Inflated Regression approach is implemented using XGBoost:

- **Classifier:** An XGBClassifier predicts whether a loan will have a non-zero recovery. This step separates the zeros from the non-zero observations.

- **Regressor:** An XGBRegressor is trained only on the non-zero recovery rates. This allows the model to focus on predicting the magnitude of recovery for cases where recovery is expected.

```python
X_train = df_train_encoded.drop(columns = 'RR')
X_test = df_test_encoded.drop(columns = 'RR')

y_train = df_train_encoded['RR']
y_test = df_test_encoded['RR']
```

```python
classifier = XGBClassifier(random_state=99,
                           objective="binary:logistic",
                           n_estimators=100,
                           max_depth=4,)
classifier.fit(X_train, y_train != 0.0)
```

```python
classifier.classes_
```

```python
regressor = XGBRegressor(random_state=99,
                         objective="reg:squarederror",
                         n_estimators=200,
                         max_depth=5,)
non_zero_mask = y_train != 0.0
regressor.fit(X_train[non_zero_mask], y_train[non_zero_mask])
```

**Combining Predictions:**

- For each observation, if the classifier predicts a non-zero recovery, the regressor’s output is used.

- If the classifier predicts zero recovery, the final prediction is set to 0.

- All predictions are clipped to the valid range [0, 1] to ensure meaningful recovery rates.

This approach effectively separates the two problems—predicting whether a recovery occurs and predicting its amount—allowing the model to better capture the underlying patterns in the highly skewed target variable.

No hyperparameter tuning is performed in this analysis; default XGBoost settings are used for simplicity and reproducibility.

```python
# 1. Predict if recovery will be zero or not
will_recover_train = classifier.predict_proba(X_train)[:, 1] > 0.5

# 2. Predict recovery amount for all train cases
recovery_amounts_train = regressor.predict(X_train)

# 3. Combine: use recovery amount only if non-zero predicted
train_predictions = np.where(will_recover_train, recovery_amounts_train, 0.0)
```

```python
y_pred_train = np.clip(train_predictions, 0, 1)
```

```python
# 1. Predict if recovery will be zero or not
will_recover_test = classifier.predict_proba(X_test)[:, 1] > 0.5

# 2. Predict recovery amount for all test cases
recovery_amounts_test = regressor.predict(X_test)

# 3. Combine: use recovery amount only if non-zero predicted
test_predictions = np.where(will_recover_test, recovery_amounts_test, 0.0)
```

```python
y_pred_test = np.clip(test_predictions, 0, 1)
```

## **Assessing model performance**

Root Mean Squared Error (RMSE) and Mean Absolute Error (MAE) are calculated on both training and test sets to quantify prediction accuracy. These metrics provide complementary insights: RMSE penalizes larger errors more heavily, while MAE gives a straightforward average of absolute differences.

Histograms comparing predicted vs actual recovery rates allow us to visually assess how well the model captures the distribution of recovery rates.

Additionally, residual distribution plots show the spread of prediction errors for both training and test sets. Histograms highlight where predictions tend to under- or overestimate the actual recovery rates, helping to identify potential biases or areas for improvement.

```python
# Training set
rmse_train = (mean_squared_error(y_train, y_pred_train))**0.5
mae_train = mean_absolute_error(y_train, y_pred_train)

# Test set
rmse_test = (mean_squared_error(y_test, y_pred_test))**0.5
mae_test = mean_absolute_error(y_test, y_pred_test)

print(f"Train RMSE: {rmse_train:.4f}, MAE: {mae_train:.4f}")
print(f"Test  RMSE: {rmse_test:.4f}, MAE: {mae_test:.4f}")
```

The LGD model demonstrates solid predictive accuracy, with Train RMSE of 0.0808 (MAE 0.0481) and Test RMSE of 0.1004 (MAE 0.0632). The model slightly overpredicts Recovery Rates (RR) on the test set — and therefore underpredicts LGD. Although this direction is not conservative on its own, it is generally acceptable in practice because regulatory LGD adjustments (e.g., downturn LGD and MoC) shift final LGD estimates upwards to ensure sufficient conservatism.
