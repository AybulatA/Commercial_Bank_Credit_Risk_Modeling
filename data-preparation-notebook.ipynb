# **Data preprocessing notebook**

This notebook focuses on building a clean and consistent preprocessing pipeline for the Lending Club dataset (obtained from Kaggle). The goal is to transform the raw loan-level data into a structured, modeling-ready dataset suitable for estimating Expected Loss (EL) — a key risk metric used by financial institutions to quantify credit risk.

LendingClub is a U.S.-based peer-to-peer (P2P) lending platform headquartered in San Francisco. It was the first P2P lender to register its loan offerings as securities with the U.S. Securities and Exchange Commission (SEC) and the first to provide a secondary market for loan trading. Over the years, it has grown into the world’s largest P2P lending platform, making its publicly available loan-level data a valuable resource for credit-risk research and model development.

We will not use all variables provided in the dataset; instead, we focus on a curated subset of features that are most relevant for credit-risk modeling and Expected Loss estimation.

The selected variables are:

`id` — Unique loan identifier.

`member_id` — Identifier of the borrower requesting the loan.

`loan_amnt` — Amount originally requested by the borrower.

`funded_amnt` — Total amount funded for the loan at that point in time.

`term` — Loan maturity in months (36 or 60).

`int_rate` — Interest rate assigned to the loan.

`installment` — Monthly payment owed by the borrower if the loan is funded.

`grade` — Credit grade assigned by LendingClub’s internal rating system.

`emp_length` — Borrower’s employment length (in years).

`home_ownership` — Borrower's home-ownership status at application.

`annual_inc` — Self-reported annual income.

`issue_d` — Month when the loan was issued.

`loan_status` — Current loan status (used to derive our target variable).

`purpose` — Borrower-provided purpose for the loan.

`addr_state` — U.S. state of residence provided in the application.

`dti` — Debt-to-income ratio, excluding mortgage and requested loan.

`delinq_2yrs` — Number of 30+ day delinquencies in the past 2 years.

`earliest_cr_line` — Date of the borrower’s earliest reported credit line.

`verification_status` — Whether income was verified by LendingClub.

`initial_list_status` — Initial listing status of the loan (W/F).

`inq_last_6mths` — Number of credit inquiries in the last 6 months.

`open_acc` — Number of currently open credit accounts.

`pub_rec` — Number of derogatory public records.

`total_acc` — Total number of credit accounts on file.

`acc_now_delinq` — Number of accounts currently delinquent.

`total_rev_hi_lim` — Total revolving high credit limit.

`recoveries` — The amount the lender managed to recover after a borrower defaulted (“post charge-off gross recovery”).

`total_rec_prncp` — The total principal amount that has been repaid to date on the loan.

```python
selected_cols = [
    "id", "member_id", "loan_amnt", "funded_amnt", "term", "int_rate", "installment", "grade", "emp_length",
    "home_ownership", "annual_inc", "issue_d", "loan_status", "purpose", "addr_state", "dti", "delinq_2yrs",
    "earliest_cr_line", "verification_status", "initial_list_status", "inq_last_6mths", "open_acc", "pub_rec",
    "total_acc", "acc_now_delinq", "total_rev_hi_lim", 'recoveries', 'total_rec_prncp'
]
df = df.loc[:, selected_cols]
df.head()
```

Before modeling, several columns require attention due to unusable formats, inconsistent data types, or information that is not relevant for our analysis.
Therefore, the first step is to clean and normalize the dataset, ensuring that all selected variables are structured, interpretable, and ready for downstream processing.

```python
df["emp_length"].value_counts()
```

```python
df.loc[~df["emp_length"].isnull(), "emp_length"] = (
    df.loc[
        ~df["emp_length"].isnull(), "emp_length"
    ].astype(str)
     .str.extract(r"([^(?:years?)]*)").iloc[:, 0]
     .str.replace(r"10\+", str(10), regex=True)
     .str.replace(r"< 1", str(0), regex=True)
     .astype(int)
)
```

```python
df["emp_length"].value_counts()
```

```python
df["term"].value_counts()
```

```python
df.loc[:, "term"] = (
    df.loc[
        :, "term"
    ].astype(str)
     .str.extract(r"(\d+)").iloc[:, 0]
     .astype(int)
)
```

```python
df["term"].value_counts()
```

```python
df["issue_d"].value_counts()
```

```python
df["issue_d"] = pd.to_datetime(df["issue_d"], format="%b-%Y")
```

```python
df["issue_d"].value_counts()
```

To make the `issue_d` field more useful for modeling, we convert it from a raw datetime into a numerical feature representing the number of months since the loan’s issue date (using December 2017 as the reference point).

```python
reference_date = datetime(2017, 12, 1)

df["months_since_issue_date"] = (reference_date - df["issue_d"]).dt.days / 30.44
```

```python
df["months_since_issue_date"].describe().round()
```

```python
df["earliest_cr_line"].value_counts().head()
```

```python
df["earliest_cr_line"] = pd.to_datetime(df["earliest_cr_line"], format="%b-%Y")
df["earliest_cr_line"].value_counts().head()
```

```python
df["months_since_earliest_cr_line"] = (reference_date - df["earliest_cr_line"]).dt.days / 30.44
```

```python
df["months_since_earliest_cr_line"].describe()
```
